```python
import pandas as pd
from pyxlsb import open_workbook
from openpyxl import load_workbook

print("Libraries Installed Successfully")
```

```python
input_file = r"C:\Users\cchara01\Downloads\Coach Spring 26 Internal Plan - April.xlsb"

output_file = "Coach_Spring_26_Internal_Plan_Converted.xlsx"
```

```python
all_sheets = {}

with open_workbook(input_file) as wb:

    for sheet_name in wb.sheets:

        print(f"Reading Sheet: {sheet_name}")

        with wb.get_sheet(sheet_name) as sheet:

            data = []

            for row in sheet.rows():

                row_data = [item.v for item in row]

                data.append(row_data)

            if len(data) > 0:

                # First row as columns
                columns = data[0]

                # Remaining rows as data
                df = pd.DataFrame(
                    data[1:],
                    columns=columns
                )

                all_sheets[sheet_name] = df

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    for sheet_name, df in all_sheets.items():

        print(f"Writing Sheet: {sheet_name}")

        # Excel sheet names max length = 31
        clean_sheet_name = str(sheet_name)[:31]

        df.to_excel(
            writer,
            sheet_name=clean_sheet_name,
            index=False
        )

print("Conversion Completed")

wb = load_workbook(output_file)

for ws in wb.worksheets:

    # Unhide sheet
    ws.sheet_state = "visible"

    # Unhide rows
    for row_dim in ws.row_dimensions.values():
        row_dim.hidden = False

    # Unhide columns
    for col_dim in ws.column_dimensions.values():
        col_dim.hidden = False

wb.save(output_file)

print("All sheets/rows/columns unhidden")

print("Converted File Saved As:")
print(output_file)

```
